```python
import pandas as pd
import numpy as np
import re
from tqdm import tqdm
from urllib.parse import urlparse

def net_loc(url):
    # Parse the URL
    try:
        parsed_url = urlparse(url)
        cleaned_url = parsed_url.netloc
    except:
    # Remove fragments
        cleaned_url = None

    return cleaned_url

df = pd.read_csv('../../data/00-newspaper_data/jr_code/merged_final_cleaned.csv')

df = df[['link', 'Column4', 'Column3', 'Column5', 'TextColumn']]
df = df.drop_duplicates('link').reset_index(drop = True)

df['net_loc'] = df['link'].apply(net_loc)

df
```

```python
links = df.drop_duplicates('net_loc')
links = links[['net_loc']]
links.to_excel('../../data/00-newspaper_data/jr_code/newspapers_links.xlsx', 
               index = False)
```
