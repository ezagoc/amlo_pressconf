```python
import pandas as pd
import numpy as np


path_24 = "../../../data/00-newspaper_data/crawler/horas24"
path_jornada = "../../../data/00-newspaper_data/crawler/jornada"
```

```python
# --- 24 Horas CDMX: merge 3 files, deduplicate, filter 2018-2024 ---
cdmx1 = pd.read_parquet(f"{path_24}/articles_cdmx.parquet")
cdmx2 = pd.read_parquet(f"{path_24}/articles_cdmx2.parquet")
cdmx3 = pd.read_parquet(f"{path_24}/articles_cdmx3.parquet")

print(f"cdmx1: {len(cdmx1):,}  cdmx2: {len(cdmx2):,}  cdmx3: {len(cdmx3):,}")

cdmx = pd.concat([cdmx1, cdmx2, cdmx3], ignore_index=True)
print(f"After concat: {len(cdmx):,}")

cdmx = cdmx.drop_duplicates(subset="url")
print(f"After dedup (url): {len(cdmx):,}")

# Normalize date to UTC-naive for filtering
cdmx["date"] = pd.to_datetime(cdmx["date"], utc=True).dt.tz_localize(None)
cdmx = cdmx[(cdmx["date"] >= "2018-01-01") & (cdmx["date"] < "2025-01-01")].reset_index(drop=True)
print(f"After 2018-2024 filter: {len(cdmx):,}")
print(f"Date range: {cdmx['date'].min()} -> {cdmx['date'].max()}")
cdmx.head(3)
```

```python
cdmx.to_parquet(f"{path_24}/articles_cdmx_final.parquet", index=False)
print(f"Saved articles_cdmx_final.parquet — {len(cdmx):,} rows")
```

```python
# --- 24 Horas Puebla: filter 2018-2024 ---
puebla = pd.read_parquet(f"{path_24}/articles_puebla.parquet")
print(f"Puebla raw: {len(puebla):,}")

puebla["date"] = pd.to_datetime(puebla["date"], utc=True).dt.tz_localize(None)
puebla = puebla[(puebla["date"] >= "2018-01-01") & (puebla["date"] < "2025-01-01")].reset_index(drop=True)
print(f"After 2018-2024 filter: {len(puebla):,}")
print(f"Date range: {puebla['date'].min()} -> {puebla['date'].max()}")
puebla.head(3)
```

```python
puebla.to_parquet(f"{path_24}/articles_puebla_final.parquet", index=False)
print(f"Saved articles_puebla_final.parquet — {len(puebla):,} rows")
```

```python
# --- Jornada: filter 2018-2024 ---
jornada = pd.read_parquet(f"{path_jornada}/articles.parquet")
print(f"Jornada raw: {len(jornada):,}")

jornada["date"] = pd.to_datetime(jornada["date"])
jornada = jornada[(jornada["date"] >= "2018-01-01") & (jornada["date"] < "2025-01-01")].reset_index(drop=True)
print(f"After 2018-2024 filter: {len(jornada):,}")
print(f"Date range: {jornada['date'].min()} -> {jornada['date'].max()}")
jornada.head(3)
```

```python
jornada.to_parquet(f"{path_jornada}/articles_final.parquet", index=False)
print(f"Saved articles_final.parquet — {len(jornada):,} rows")
```
