```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

def extract_all_urls(base_url):
    """
    Extract all URLs from the given webpage.
    Args:
        base_url (str): The URL of the webpage to scrape.
    Returns:
        list: A list of full URLs found on the webpage.
    """
    try:
        # Send an HTTP request to the page
        response = requests.get(base_url, timeout=5)
        response.raise_for_status()
        
        # Parse the HTML content
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # Collect all URLs from <a> tags
        urls = set()  # Use a set to avoid duplicates
        for link in soup.find_all("a", href=True):
            # Convert relative URLs to absolute URLs
            full_url = urljoin(base_url, link['href'])
            urls.add(full_url)
        
        return list(urls)
    
    except requests.RequestException as e:
        print(f"Error fetching URLs from {base_url}: {e}")
        return []

# Example usage
base_url = "https://atiempo.tv/category/elecciones-2024/page/5/"  # Replace with your target URL
urls = extract_all_urls(base_url)
print("Extracted URLs:")
for url in urls:
    print(url)
```
