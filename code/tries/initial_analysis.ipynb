```python
import os
import pandas as pd
from datetime import datetime

def import_files_in_time_period(base_path, start_date, end_date, file_name=None, date_format="%d %B, %Y"):
    """
    Imports files from a nested folder structure and filters by a specified time period.

    Args:
        base_path (str): The root path containing year/month-year/month day, year folders.
        start_date (str): Start date for filtering (in 'YYYY-MM-DD' format).
        end_date (str): End date for filtering (in 'YYYY-MM-DD' format).
        file_name (str): Name of the specific file to import. If None, imports all files in the range.
        date_format (str): Format of the date in folder names (e.g., '%d %B, %Y' for '17 febrero, 2021').

    Returns:
        list of pandas.DataFrame: List of DataFrames for the files imported.
    """
    imported_data = []
    start_date = datetime.strptime(start_date, "%Y-%m-%d")
    end_date = datetime.strptime(end_date, "%Y-%m-%d")

    # Walk through the directory structure
    for year_folder in os.listdir(base_path):
        year_path = os.path.join(base_path, year_folder)
        if not os.path.isdir(year_path):
            continue

        for month_folder in os.listdir(year_path):
            month_path = os.path.join(year_path, month_folder)
            if not os.path.isdir(month_path):
                continue

            for day_folder in os.listdir(month_path):
                day_path = os.path.join(month_path, day_folder)
                if not os.path.isdir(day_path):
                    continue

                # Parse the date from the folder name
                try:
                    folder_date = datetime.strptime(day_folder, date_format)
                except ValueError:
                    print(f"Skipping folder with invalid date format: {day_folder}")
                    continue

                # Check if the folder date falls within the time period
                if start_date <= folder_date <= end_date:
                    # If a specific file name is given, look for it
                    if file_name:
                        file_path = os.path.join(day_path, file_name)
                        if os.path.exists(file_path):
                            imported_data.append(pd.read_csv(file_path))
                    else:
                        # Import all files in the folder if no specific file name is given
                        for file in os.listdir(day_path):
                            file_path = os.path.join(day_path, file)
                            if os.path.isfile(file_path):
                                imported_data.append(pd.read_csv(file_path))
    
    return imported_data

# Example Usage:
# base_path = "/path/to/data"
# start_date = "2021-01-01"
# end_date = "2021-12-31"
# file_name = "data.csv"  # Optional
# dfs = import_files_in_time_period(base_path, start_date, end_date, file_name, "%d %B, %Y")
# for df in dfs:
#     print(df.head())

```
